```python
from IPython.core.interactiveshell import InteractiveShell
InteractiveShell.ast_node_interactivity = "all"
import numpy as np
import pandas as pd
```

```python
orioles_df = pd.read_csv('Orioles_Hitting_2025.csv', index_col='Player')
rays_df = pd.read_csv('Rays_Hitting_2025.csv', index_col='Player')
redsox_df = pd.read_csv('RedSox_Hitting_2025.csv', index_col='Player')
bluejays_df = pd.read_csv('BlueJays_Hitting_2025.csv', index_col='Player')

print("Orioles Roster:")
print(orioles_df)
print(f"\nOrioles hitters: {len(orioles_df)}")

print("\nRays Roster:")
print(rays_df)
print(f"\nRays hitters: {len(rays_df)}")

print("\nRed Sox Roster:")
print(redsox_df)
print(f"\nRed Sox hitters: {len(redsox_df)}")

print("\nBlue Jays Roster:")
print(bluejays_df)
print(f"\nBlue Jays hitters: {len(bluejays_df)}")
```

```python
all_teams_df = pd.concat([orioles_df, rays_df, redsox_df, bluejays_df])

print("\nCombined Roster:")
print(all_teams_df)
print(f"\nTotal hitters: {len(all_teams_df)}")
```

```python
all_teams_df.to_csv('AL_East_No_Yankees_2025.csv', index=True)

print("\nCSV file 'AL_East_No_Yankees_2025.csv'")
```
